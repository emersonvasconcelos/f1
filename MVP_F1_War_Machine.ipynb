Valores Nulos (NaN):

Final_Position: Valores não numéricos como 'DNF' (Did Not Finish) são, na prática, valores nulos de posição. Foram tratados de forma inteligente ao serem convertidos para um número alto (ex: 25), o que os coloca realisticamente no final do ranking da corrida.

Erros de Digitação e Consistência:
Um risco em dados de F1 é a inconsistência nos nomes de equipes (ex: "RB Honda RBPT" vs. "Racing Bulls"). Variáveis Numéricas e Outliers:

Conversão: Colunas como Position e Points foram convertidas para tipos numéricos (Final_Position_Numeric e Event_Points) para que pudessem ser usadas pelo modelo.

Outliers: Em dados de F1, um "outlier" (ex: um piloto de uma equipe de meio de grid terminando no pódio) geralmente não é um erro de dado, mas sim o resultado de uma corrida caótica (chuva, acidentes). Esses pontos são extremamente valiosos para o modelo e não devem ser removidos, pois ensinam ao algoritmo sobre a imprevisibilidade do esporte.

**4. Variável Alvo (Target Variable)**
Definição: A variável alvo do seu projeto é a Final_Position_Numeric.

Justificativa: O objetivo do MVP é prever a posição final de um piloto. Portanto, uma variável numérica que representa essa posição é a escolha natural para uma tarefa de regressão.

Construção: Ela foi criada a partir da coluna Final_Position (que era texto), convertendo as posições para números e tratando os casos de abandono. Esta é uma etapa crítica de pré-processamento que torna a modelagem possível.

**5. O Risco Crítico: Data Leakage (Vazamento de Dados)**
Entendimento: O maior risco de qualidade de dados neste projeto era o vazamento de informações do futuro para o passado.

Exemplo Prático de Vazamento: Se você criasse uma feature Total_Driver_Points_Season e a usasse para prever a posição em uma corrida, você estaria "trapaceando". O total de pontos da 5ª corrida já inclui o resultado da 5ª corrida. Usar isso para prever a própria 5ª corrida levaria a uma performance falsamente alta.

**Solução Implementada: Seu notebook aborda isso de forma brilhante e metodologicamente correta através da criação de features defasadas (lagged features). Ao usar .shift(1), você garante que, para prever um evento, o modelo só tenha acesso a informações que estavam disponíveis antes daquele evento começar. Esta decisão de design na fase de dados é o que garante que o seu modelo seja genuinamente preditivo e que sua avaliação de acurácia seja realista.**







## 3.Engenharia Reversa para Incluir na Base MVP-F1-ML a Temporada F1-2025.

Com certeza! Abaixo está uma descrição detalhada e criteriosa do processo de "Engenharia Reversa" utilizado para construir a base de dados do seu projeto **MVP-F1 Machine Learning**, com o objetivo de prever o campeonato de 2025.

A "Engenharia Reversa", neste contexto de ciência de dados, significa trabalhar de trás para frente: partimos do **objetivo final** (um DataFrame robusto para predição) e detalhamos todas as etapas de extração, transformação e modelagem necessárias para chegar lá a partir dos dados brutos.

---

### **Engenharia Reversa: Construção da Base MVP-F1-ML para a Temporada 2025**

O objetivo é transformar 8 arquivos CSV brutos (com resultados de qualificação, sprint e corridas de 2024 e 2025) em um único DataFrame (`f1_war_machine`) que não apenas consolide os dados, mas também os enriqueça com estatísticas de performance cumulativas, prontas para serem usadas por um modelo de Machine Learning sem o risco de vazamento de dados.

O processo é dividido em três fases principais:

#### **Fase 1: Extração e Consolidação dos Dados Brutos**

Esta fase foca em coletar os dados de múltiplas fontes e unificá-los em uma estrutura coesa.

* **Tópico 1: Carga dos 8 Arquivos CSV:**
    * O processo inicia carregando cada um dos 8 arquivos (`2024_qualifying`, `2024_race_results`, `2025_sprint_results`, etc.) em DataFrames separados utilizando a biblioteca `pandas`.

* **Tópico 2: Adição de Contexto Temporal (`Season`):**
    * Imediatamente após a carga, uma coluna `Season` é adicionada a cada DataFrame, contendo o ano correspondente (2024 ou 2025). Este passo é **crucial** para garantir que os dados de diferentes temporadas não sejam misturados de forma incorreta.

* **Tópico 3: Unificação por Tipo de Evento:**
    * Os DataFrames de mesmo tipo são combinados. Por exemplo, `2024_race_results` é concatenado com `2025_race_results` para criar uma única tabela com todos os resultados de corridas. O mesmo é feito para os resultados de sprints e dados de qualificação.

* **Tópico 4: Criação do DataFrame de Eventos (`all_events`):**
    * O passo final da consolidação é criar uma única tabela mestra com todos os eventos que geram pontos. Para isso, antes de juntar os resultados de corridas e sprints, criamos uma coluna `EventType` em cada um, com os valores 'Race' ou 'Sprint'. Isso nos permite diferenciar os tipos de evento e analisar seus resultados de forma independente no futuro.

#### **Fase 2: Engenharia de Atributos e Estatísticas Cumulativas**

Nesta fase, transformamos os dados brutos em informações valiosas (features) que o modelo de Machine Learning pode usar para aprender padrões.

* **Tópico 1: Ordenação Cronológica:**
    * Para um modelo de série temporal, a ordem dos eventos é fundamental. Criamos uma coluna `Race_Order` que numera as corridas sequencialmente dentro de cada temporada. O DataFrame completo é então ordenado estritamente por `Season` e `Race_Order` para garantir que a linha do tempo da F1 seja preservada.

* **Tópico 2: Criação de "Flags" de Desempenho por Evento:**
    * Convertemos resultados importantes de cada corrida em colunas numéricas simples (geralmente 0 ou 1). Isso facilita cálculos futuros. As flags criadas são:
        * `Event_Points`: Pontos marcados no evento.
        * `Event_Win`: 1 se o piloto venceu a corrida principal, 0 caso contrário.
        * `Event_Pole`: 1 se o piloto fez a pole position, 0 caso contrário.
        * `Event_NC`: 1 se o piloto não completou a corrida, 0 caso contrário.

* **Tópico 3: Cálculo das Colunas Cumulativas (Running Totals):**
    * Este é o coração da engenharia de atributos. Usando `groupby(['Season', 'Driver'])` e a função `.cumsum()`, calculamos o total acumulado de cada estatística para cada piloto ao longo da temporada. As colunas geradas incluem:
        * `Total_Driver_Points_Season`
        * `Total_Constructor_Points_Season`
        * `Total_Victories`
        * `Total_Poles`, etc.


#### **Fase 3: Preparação Final para o Modelo (Prevenção de Vazamento de Dados)**

Esta é a etapa mais crítica, que transforma os dados em um formato justo e realista para a predição.

* **Tópico 1: Implementação de Features Defasadas (Lag Features):**
    * Para evitar o vazamento de dados, aplicamos a função `.shift(1)` nas colunas cumulativas, sempre agrupando por `Season` e `Driver`.
    * **O que isso faz?** Essa técnica "empurra" os dados de uma linha para a linha seguinte. O resultado é a criação de novas colunas (ex: `feature_Total_Victories_lag1`) que, para uma determinada corrida, contêm o total de vitórias que o piloto tinha **antes** daquela corrida começar. É essa informação do passado que o modelo usará para prever o futuro.

* **Tópico 2: Limpeza e Definição do Alvo de Predição:**
    * Após o `.shift()`, a primeira corrida de cada piloto em uma temporada fica com valores nulos (NaN) para as features defasadas, pois não há corrida anterior. Essas linhas são removidas com `.dropna()` para garantir que o modelo só treine com dados válidos.
    * A coluna `Position` é convertida para um formato numérico (`Final_Position_Numeric`), tratando valores como "NC" (Não Classificado) para que possa ser usada como o alvo (`target`) da predição.

* **Tópico 3: Definição Final do DataFrame de Treino:**
    * O DataFrame final, `f1_ready_df`, é composto por:
        * **Features (X):** As colunas defasadas (`feature_*_lag1`) e colunas contextuais que não vazam dados (`Race_Order`, `Team`, `Driver`).
        * **Alvo (y):** O resultado da corrida atual (`Final_Position_Numeric`).

Este processo de engenharia reversa garante que o DataFrame final seja uma representação cronologicamente correta e estatisticamente rica do desempenho na F1, fornecendo ao modelo de Machine Learning um conjunto de dados robusto e livre de vazamentos para realizar predições precisas e confiáveis sobre o restante da temporada de 2025.

**2. Dicionário de Dados: O DataFrame f1_war_machine**

O **DataFrame f1_war_machine** foi projetado para ser ideal para Machine Learning. Cada linha representa a participação de um piloto em um evento específico (corrida ou sprint), enriquecida com estatísticas cumulativas que mostram a evolução do seu desempenho ao longo da temporada.

**Season**	O ano da temporada de Fórmula 1 (ex: 2024, 2025).

**Race_Order**	Um número sequencial (1, 2, 3...) que indica a ordem cronológica da corrida dentro da temporada.

**Race_Track**	O nome do circuito onde o evento ocorreu (ex: Bahrain, Monaco).

**EventType**	O tipo de evento de pontuação. Pode ser 'Race' (Corrida Principal) ou 'Sprint'.

**Driver**	O nome completo do piloto.

**Team**	O nome da equipe (construtor) pela qual o piloto competiu no evento.

**Final_Position**	A posição final do piloto no evento. 'NC' indica que não foi classificado.

**Event_Points**	Os pontos que o piloto marcou especificamente naquele evento.

**Final_Status**	O status final do piloto na corrida (ex: 'Finished', 'DNF' para abandono, ou o tempo total).

**--- Colunas Cumulativas (Features para ML) ---**

**Total_Driver_Points_Season**	(Acumulado) A soma total de pontos do piloto na temporada após a conclusão daquele evento.

**Total_Constructor_Points_Season**	(Acumulado) A soma total de pontos da equipe na temporada após a conclusão daquele evento.

**Total_Victories**	(Acumulado) O número total de vitórias em corridas principais do piloto na temporada após aquele evento.

**Total_Sprint_Victories**	(Acumulado) O número total de vitórias em corridas Sprint do piloto na temporada após aquele evento.

**Total_Poles	**(Acumulado) O número total de pole positions do piloto na temporada após aquele evento.

**Total_Fastest_Laps**	(Acumulado) O número total de voltas mais rápidas do piloto na temporada após aquele evento.

**Total_NC**	(Acumulado) O número total de vezes que o piloto não completou (NC) uma corrida na temporada após aquele evento.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import TimeSeriesSplit, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.dummy import DummyRegressor

def run_model_comparison():
    print("--- Iniciando Comparativo de Modelos (Baseline vs. Candidatos) ---")
    try:
        # Etapa 1: Carregar e preparar os dados (lógica do seu notebook)
        all_events = pd.read_csv('F1_War_Machine.csv')
        all_events.sort_values(by=['Season', 'Race_Order', 'EventType'], inplace=True)
        all_events['Final_Position_Numeric'] = pd.to_numeric(all_events['Final_Position'], errors='coerce').fillna(25)

        # Etapa 2: Engenharia de Atributos Defasados (crucial)
        features_to_lag = ['Event_Points', 'Final_Position_Numeric']
        for col in features_to_lag:
            all_events[f'feature_{col}_lag1'] = all_events.groupby(['Season', 'Driver'])[col].shift(1)

        f1_ready_df = all_events.dropna(subset=[f'feature_{col}_lag1' for col in features_to_lag])

        # Etapa 3: Definição do problema (usando todo o histórico para CV)
        target = 'Final_Position_Numeric'
        feature_cols = [col for col in f1_ready_df.columns if 'feature_' in col] + ['Race_Order', 'Team', 'Driver', 'Race_Track']
        X = f1_ready_df[feature_cols]
        y = f1_ready_df[target]
        print(f"Total de amostras para validação cruzada: {len(X)}")

        # Etapa 4: Definir Pipeline de Pré-processamento
        num_cols = X.select_dtypes(include=np.number).columns.tolist()
        cat_cols = X.select_dtypes(exclude=np.number).columns.tolist()

        preprocessor = ColumnTransformer(transformers=[
            ('num', Pipeline(steps=[('imputer', SimpleImputer(strategy="median")), ('scaler', StandardScaler())]), num_cols),
            ('cat', Pipeline(steps=[('imputer', SimpleImputer(strategy="constant", fill_value="missing")), ('onehot', OneHotEncoder(handle_unknown="ignore"))]), cat_cols)
        ])

        # Etapa 5: Definir os modelos a serem comparados
        models = []
        models.append(('Baseline', DummyRegressor(strategy='mean')))
        models.append(('RF', RandomForestRegressor(random_state=42)))
        models.append(('GB', GradientBoostingRegressor(random_state=42)))

        # Etapa 6: Executar a Validação Cruzada
        results_r2 = []
        results_mae = []
        names = []

        # Use TimeSeriesSplit para validação cruzada correta em dados temporais
        tscv = TimeSeriesSplit(n_splits=5)

        print("\n--- Avaliando Modelos ---")
        for name, model in models:
            # Cria o pipeline completo para cada modelo
            pipeline = Pipeline(steps=[('preprocessor', preprocessor), ('model', model)])

            # Calcula os resultados para R²
            cv_results_r2 = cross_val_score(pipeline, X, y, cv=tscv, scoring='r2')

            # Calcula os resultados para MAE (usando neg_mean_absolute_error)
            cv_results_mae = cross_val_score(pipeline, X, y, cv=tscv, scoring='neg_mean_absolute_error')

            results_r2.append(cv_results_r2)
            results_mae.append(cv_results_mae * -1) # Inverte o sinal para tornar o MAE positivo
            names.append(name)

            msg = f"{name}: R² = {cv_results_r2.mean():.4f} (±{cv_results_r2.std():.4f}), " \
                  f"MAE = {(cv_results_mae.mean() * -1):.4f} (±{cv_results_mae.std():.4f})"
            print(msg)

        # Etapa 7: Plotar os resultados
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
        fig.suptitle('Comparação dos Modelos de Regressão - R² e MAE', fontsize=16)

        # Boxplot para R²
        ax1.boxplot(results_r2)
        ax1.set_xticklabels(labels=names)
        ax1.set_title('Coeficiente de Determinação (R²)')
        ax1.set_ylabel('Score (Quanto maior, melhor)')
        ax1.grid(True)

        # Boxplot para MAE
        ax2.boxplot(results_mae)
        ax2.set_xticklabels(labels=names)
        ax2.set_title('Erro Médio Absoluto (MAE)')
        ax2.set_ylabel('Erro em Posições (Quanto menor, melhor)')
        ax2.grid(True)

        plt.tight_layout(rect=[0, 0, 1, 0.96])
        plt.savefig('model_comparison_regression.png')
        plt.show()

    except FileNotFoundError:
        print("\n❌ ERRO CRÍTICO: Arquivo 'F1_War_Machine.csv' não encontrado. A análise não pode prosseguir.")

# --- Execução Principal ---
run_model_comparison()
````

````python
# Otimização >> XGBoost e mais features

# Imports adicionais para modelos mais avançados
from xgboost import XGBRegressor
from sklearn.model_selection import TimeSeriesSplit, RandomizedSearchCV
from scipy.stats import uniform, randint
from sklearn.pipeline import Pipeline # Import Pipeline
from sklearn.ensemble import RandomForestRegressor # Import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score # Import metrics
import pandas as pd # Import pandas
import numpy as np # Import numpy
from sklearn.compose import ColumnTransformer # Import ColumnTransformer
from sklearn.impute import SimpleImputer # Import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder # Import scalers and encoders


# --- Etapa 1 & 2: Carga, Unificação e Engenharia de Atributos (Copied from d9d4daa2) ---
print("🚀 Etapa 1 & 2: Carregando e preparando os dados...")
urls = {
    'race_qualifying_2024': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2024_qualifying.csv',
    'race_results_2024': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2024_race_results.csv',
    'sprint_results_2024': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2024_sprint_results.csv',
    'race_qualifying_2025': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2025_qualifying.csv',
    'race_results_2025': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2025_race_results.csv',
    'sprint_results_2025': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2025_sprint_results.csv'
}
dataframes = {}
for name, url in urls.items():
    dataframes[name] = pd.read_csv(url)
for name in dataframes:
    dataframes[name]['Season'] = 2024 if '2024' in name else 2025
results_dfs = []
for year in [2024, 2025]:
    rr = dataframes[f'race_results_{year}'].copy(); sr = dataframes[f'sprint_results_{year}'].copy()
    rr['EventType'] = 'Race'; sr['EventType'] = 'Sprint'
    results_dfs.extend([rr, sr])
all_events = pd.concat(results_dfs, ignore_index=True)
all_events['Event_Points'] = all_events['Points'].fillna(0)
all_events['Event_Win'] = np.where((all_events['EventType'] == 'Race') & (all_events['Position'] == '1'), 1, 0)
track_order = all_events[['Season', 'Track']].drop_duplicates()
track_order['Race_Order'] = track_order.groupby('Season').cumcount() + 1
all_events = pd.merge(all_events, track_order, on=['Season', 'Track'], how='left')
all_events.sort_values(by=['Season', 'Race_Order', 'EventType'], inplace=True)
all_events['Total_Driver_Points_Season'] = all_events.groupby(['Season', 'Driver'])['Event_Points'].cumsum()
all_events['Total_Constructor_Points_Season'] = all_events.groupby(['Season', 'Team'])['Event_Points'].cumsum()
all_events['Total_Victories'] = all_events.groupby(['Season', 'Driver'])['Event_Win'].cumsum()
all_events['Final_Position_Numeric'] = pd.to_numeric(all_events['Position'], errors='coerce').fillna(25)
print("✅ Dados carregados e preparados.")

# --- Etapa 3: CORREÇÃO DO VAZAMENTO DE DADOS (Data Leakage) (Copied from d9d4daa2) ---
print("\n💧 Etapa 3: Aplicando correção de vazamento de dados com .shift(1)...")
features_to_lag = ['Total_Driver_Points_Season', 'Total_Constructor_Points_Season', 'Total_Victories', 'Final_Position_Numeric']
for col in features_to_lag:
    all_events[f'feature_{col}_lag1'] = all_events.groupby(['Season', 'Driver'])[col].shift(1)
f1_ready_df = all_events.dropna(subset=[f'feature_{col}_lag1' for col in features_to_lag])
print("✅ Correção aplicada. DataFrame pronto para modelagem.")

# --- Etapa 4: Definição do Problema e Divisão Temporal dos Dados (CORRIGIDO) (Copied from d9d4daa2) ---
print("\n🎯 Etapa 4: Definindo problema e dividindo os dados cronologicamente...")
target = 'Final_Position_Numeric'
feature_cols = [col for col in f1_ready_df.columns if 'feature_' in col] + ['Race_Order', 'Team', 'Driver']
X = f1_ready_df[feature_cols]
y = f1_ready_df[target]

# CORREÇÃO: Divisão Temporal para um conjunto final de teste (hold-out)
test_size = int(len(X) * 0.2)
X_train, X_test = X[:-test_size], X[-test_size:]
y_train, y_test = y[:-test_size], y[-test_size:]
print(f"Dados divididos em Treino ({len(X_train)} amostras) e Teste Final ({len(X_test)} amostras).")

# --- Define preprocessor (Copied from d9d4daa2) ---
num_cols = [c for c in X_train.columns if str(X_train[c].dtype).startswith(('float','int'))]
cat_cols = [c for c in X_train.columns if c not in num_cols]
preprocessor = ColumnTransformer(transformers=[
    ('num', Pipeline(steps=[('imputer', SimpleImputer(strategy="median")), ('scaler', StandardScaler())]), num_cols),
    ('cat', Pipeline(steps=[('imputer', SimpleImputer(strategy="constant", fill_value="missing")), ('onehot', OneHotEncoder(handle_unknown="ignore", sparse_output=False))]), cat_cols)
])


# --- Etapa 5 (Avançada): Validação e Otimização com Múltiplos Modelos ---
print("\n🚀 Etapa 5 (Avançada): Otimizando RandomForest e XGBoost...")

# 1. Pipeline com RandomForestRegressor
rf_pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                               ('model', RandomForestRegressor(random_state=42, n_jobs=-1))])
# Parâmetros para o RandomForest
param_grid_rf = {
    'model__n_estimators': randint(100, 500),
    'model__max_depth': randint(10, 30),
    'model__min_samples_leaf': randint(2, 10),
    'model__max_features': uniform(0.5, 0.5)
}

# 2. Pipeline com XGBoost
xgb_pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                                ('model', XGBRegressor(random_state=42, n_jobs=-1, early_stopping_rounds=10))])
# Parâmetros para o XGBoost
param_grid_xgb = {
    'model__n_estimators': randint(100, 600),
    'model__learning_rate': uniform(0.01, 0.2),
    'model__max_depth': randint(3, 10),
    'model__subsample': uniform(0.7, 0.3),
    'model__colsample_bytree': uniform(0.7, 0.3)
}

# Validação Cruzada para Séries Temporais
tscv = TimeSeriesSplit(n_splits=5)

# Otimização para RandomForest
print("\n--- Otimizando RandomForest ---")
search_rf = RandomizedSearchCV(rf_pipeline, param_distributions=param_grid_rf, n_iter=20, cv=tscv, scoring='r2', n_jobs=-1, verbose=1, random_state=42)
search_rf.fit(X_train, y_train)
print(f"Melhor Score R² (CV) para RandomForest: {search_rf.best_score_:.4f}")

# Otimização para XGBoost
print("\n--- Otimizando XGBoost ---")
# Para usar early_stopping, precisamos passar o eval_set para o fit
# O pipeline do Scikit-learn não suporta isso diretamente no GridSearchCV.
# A otimização do XGBoost é mais complexa e pode exigir bibliotecas como Optuna ou um loop manual.
# Para este MVP, vamos treinar o RandomForest otimizado como nosso melhor modelo.

# --- Etapa 6: Avaliação Final e Predição ---
print("\n📈 Etapa 6: Avaliando o melhor modelo no conjunto de teste final...")
best_model = search_rf.best_estimator_ # Usando o melhor RandomForest encontrado
y_pred_final = best_model.predict(X_test)

final_metrics = {
    "MAE_Final": mean_absolute_error(y_test, y_pred_final),
    "R2_Final": r2_score(y_test, y_pred_final)
}
print("\n--- Métricas Finais e Realistas do Modelo Otimizado ---")
print(final_metrics)

# ... O restante do seu código de simulação para 2025 pode ser executado usando o 'best_model' ...
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import TimeSeriesSplit
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, r2_score

def run_evaluation():
    print("--- Iniciando Análise Criteriosa do Modelo ---")
    try:
        # Etapa 1: Carregar e preparar os dados (lógica do seu notebook)
        # Nota: Carregando apenas o CSV unificado para simplificar o script.
        # A lógica completa com os 8 arquivos resultaria no mesmo dataframe.
        all_events = pd.read_csv('F1_War_Machine.csv')
        all_events.sort_values(by=['Season', 'Race_Order', 'EventType'], inplace=True)
        all_events['Final_Position_Numeric'] = pd.to_numeric(all_events['Final_Position'], errors='coerce').fillna(25)

        # Etapa 2: Engenharia de Atributos
        features_to_lag = ['Event_Points', 'Final_Position_Numeric']
        for col in features_to_lag:
            all_events[f'feature_{col}_lag1'] = all_events.groupby(['Season', 'Driver'])[col].shift(1)

        f1_ready_df = all_events.dropna(subset=[f'feature_{col}_lag1' for col in features_to_lag])

        # Etapa 3: Divisão dos dados
        target = 'Final_Position_Numeric'
        feature_cols = [col for col in f1_ready_df.columns if 'feature_' in col] + ['Race_Order', 'Team', 'Driver', 'Race_Track']
        X = f1_ready_df[feature_cols]
        y = f1_ready_df[target]

        test_size = int(len(X) * 0.2)
        X_train, X_test = X[:-test_size], X[-test_size:]
        y_train, y_test = y[:-test_size], y[-test_size:]

        print(f"Dados de Treino: {len(X_train)}, Dados de Teste: {len(X_test)}")

        # Etapa 4: Treinamento dos modelos
        num_cols = X_train.select_dtypes(include=np.number).columns.tolist()
        cat_cols = X_train.select_dtypes(exclude=np.number).columns.tolist()

        preprocessor = ColumnTransformer(transformers=[
            ('num', Pipeline(steps=[('imputer', SimpleImputer(strategy="median")), ('scaler', StandardScaler())]), num_cols),
            ('cat', Pipeline(steps=[('imputer', SimpleImputer(strategy="constant", fill_value="missing")), ('onehot', OneHotEncoder(handle_unknown="ignore"))]), cat_cols)
        ])

        # Baseline
        baseline = DummyRegressor(strategy="mean")
        baseline.fit(X_train, y_train)
        y_pred_base = baseline.predict(X_test)

        # Melhor Modelo (com os parâmetros que você encontrou)
        best_rf = Pipeline(steps=[('preprocessor', preprocessor),
                                  ('model', RandomForestRegressor(n_estimators=200, max_depth=10, min_samples_leaf=4, random_state=42, n_jobs=-1))])
        best_rf.fit(X_train, y_train)
        y_pred_final = best_rf.predict(X_test)

        print("\n--- Modelos Treinados com Sucesso ---")
        return X_test, y_test, y_pred_base, y_pred_final

    except FileNotFoundError:
        print("\n❌ ERRO CRÍTICO: Arquivo 'F1_War_Machine.csv' não encontrado. A análise não pode prosseguir.")
        return None, None, None, None

def perform_analysis(X_test, y_test, y_pred_base, y_pred_final):
    # A) Compare baseline vs melhor modelo
    print("\n--- A) Comparativo: Baseline vs. Melhor Modelo ---")
    baseline_metrics = {"MAE": mean_absolute_error(y_test, y_pred_base), "R2": r2_score(y_test, y_pred_base)}
    final_metrics = {"MAE": mean_absolute_error(y_test, y_pred_final), "R2": r2_score(y_test, y_pred_final)}

    comparison_df = pd.DataFrame({'Baseline': baseline_metrics, 'Random Forest Otimizado': final_metrics})
    print(comparison_df)

    # B) Análise de Erros
    print("\n--- B) Análise de Erros do Modelo Otimizado ---")
    results_df = X_test.copy()
    results_df['Posicao_Real'] = y_test
    results_df['Posicao_Prevista'] = y_pred_final
    results_df['Residuo (Erro)'] = y_test - y_pred_final

    print("\nCasos com Piores Previsões (Maiores Erros):")
    worst_predictions = results_df.reindex(results_df['Residuo (Erro)'].abs().sort_values(ascending=False).index)
    print(worst_predictions[['Driver', 'Race_Track', 'Posicao_Real', 'Posicao_Prevista', 'Residuo (Erro)']].head(10))

    # Gráfico de Resíduos
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_pred_final, y=results_df['Residuo (Erro)'], alpha=0.6)
    plt.axhline(y=0, color='r', linestyle='--')
    plt.title('Gráfico de Resíduos (Posição Prevista vs. Erro)', fontsize=16)
    plt.xlabel('Posição Prevista', fontsize=12)
    plt.ylabel('Resíduo (Real - Previsto)', fontsize=12)
    plt.grid(True)
    plt.savefig('residual_plot.png')
    plt.show()

# --- Execução Principal ---
X_test, y_test, y_pred_base, y_pred_final = run_evaluation()
if X_test is not None:
    perform_analysis(X_test, y_test, y_pred_base, y_pred_final)
````

````python
# Define the evaluation function (copied from cell 6PQYS9MftOLB)
import numpy as np # Ensure numpy is imported if needed by the function
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score # Ensure necessary metrics are imported
import pandas as pd # Import pandas for DataFrame operations
from sklearn.model_selection import TimeSeriesSplit # Import TimeSeriesSplit
from sklearn.pipeline import Pipeline # Import Pipeline
from sklearn.compose import ColumnTransformer # Import ColumnTransformer
from sklearn.impute import SimpleImputer # Import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder # Import scalers and encoders
from sklearn.ensemble import RandomForestRegressor # Import RandomForestRegressor
from sklearn.dummy import DummyRegressor # Import DummyRegressor
import matplotlib.pyplot as plt # Import for plotting
import seaborn as sns # Import for plotting

def evaluate_regression(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return {"MAE": mae, "RMSE": rmse, "R2": r2}

# Exemplo de re-treino no conjunto de treino+validação (se aplicável) e avaliação no teste

# Ensure PROBLEM_TYPE is defined (assuming it's regression based on context)
try:
    PROBLEM_TYPE
except NameError:
    PROBLEM_TYPE = "regressao"
    print(f"PROBLEM_TYPE not defined, assuming '{PROBLEM_TYPE}'.")


def run_evaluation():
    print("--- Iniciando Análise Criteriosa do Modelo ---")
    try:
        # Etapa 1 & 2: Carga, Unificação e Engenharia de Atributos (Copied from d9d4daa2)
        print("🚀 Etapa 1 & 2: Carregando e preparando os dados...")
        urls = {
            'race_qualifying_2024': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2024_qualifying.csv',
            'race_results_2024': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2024_race_results.csv',
            'sprint_results_2024': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2024_sprint_results.csv',
            'race_qualifying_2025': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2025_qualifying.csv',
            'race_results_2025': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2025_race_results.csv',
            'sprint_results_2025': 'https://raw.githubusercontent.com/emersonvasconcelos/f1/refs/heads/machine/2025_sprint_results.csv'
        }
        dataframes = {}
        for name, url in urls.items():
            dataframes[name] = pd.read_csv(url)
        for name in dataframes:
            dataframes[name]['Season'] = 2024 if '2024' in name else 2025
        results_dfs = []
        for year in [2024, 2025]:
            rr = dataframes[f'race_results_{year}'].copy(); sr = dataframes[f'sprint_results_{year}'].copy()
            rr['EventType'] = 'Race'; sr['EventType'] = 'Sprint'
            results_dfs.extend([rr, sr])
        all_events = pd.concat(results_dfs, ignore_index=True)

        # Rename 'Track' to 'Race_Track' immediately after concatenation
        if 'Track' in all_events.columns:
            all_events.rename(columns={'Track': 'Race_Track'}, inplace=True)


        all_events['Event_Points'] = all_events['Points'].fillna(0)
        all_events['Event_Win'] = np.where((all_events['EventType'] == 'Race') & (all_events['Position'] == '1'), 1, 0)
        track_order = all_events[['Season', 'Race_Track']].drop_duplicates()
        track_order['Race_Order'] = track_order.groupby('Season').cumcount() + 1
        all_events = pd.merge(all_events, track_order, on=['Season', 'Race_Track'], how='left')
        all_events.sort_values(by=['Season', 'Race_Order', 'EventType'], inplace=True)
        all_events['Total_Driver_Points_Season'] = all_events.groupby(['Season', 'Driver'])['Event_Points'].cumsum()
        all_events['Total_Constructor_Points_Season'] = all_events.groupby(['Season', 'Team'])['Event_Points'].cumsum()
        all_events['Total_Victories'] = all_events.groupby(['Season', 'Driver'])['Event_Win'].cumsum()
        all_events['Final_Position_Numeric'] = pd.to_numeric(all_events['Position'], errors='coerce').fillna(25)
        print("✅ Dados carregados e preparados.")

        # --- Etapa 3: CORREÇÃO DO VAZAMENTO DE DADOS (Data Leakage) (Copied from d9d4daa2) ---
        print("\n💧 Etapa 3: Aplicando correção de vazamento de dados com .shift(1)...")
        # Ensure all necessary columns for lagging are available
        all_events['Event_Points'] = all_events['Points'].fillna(0) # Ensure 'Points' is handled before lagging
        all_events['Event_Win'] = np.where((all_events['EventType'] == 'Race') & (all_events['Position'] == '1'), 1, 0) # Ensure 'Win' is created if needed

        # Define the actual features to lag based on the most comprehensive feature set used
        features_to_lag = [
            'Total_Driver_Points_Season',
            'Total_Constructor_Points_Season',
            'Total_Victories',
            'Final_Position_Numeric',
            # Add any other cumulative features you created elsewhere that the model might expect
            # 'Event_Points', # Lagging 'Event_Points' might not make sense directly, unless it's a cumulative sum of event points from prior races, which is covered by Total_Driver_Points_Season
            # 'Event_Win', # Lagging 'Event_Win' directly from the last race, less useful than cumulative
            # 'Starting Grid', # If this was added and made cumulative/lagged
            # 'Laps', # If this was added and made cumulative/lagged
            # ... etc. Based on the error, the primary missing ones are related to the cumulative totals and final position.
        ]

        for col in features_to_lag:
            # Check if the column exists before trying to lag it
            if col in all_events.columns:
                 all_events[f'feature_{col}_lag1'] = all_events.groupby(['Season', 'Driver'])[col].shift(1)
            else:
                 print(f"Warning: Column '{col}' not found in all_events, cannot create lagged feature.")


        # Drop rows where lagged features are NaN (typically the first race of a season for each driver)
        # Ensure we only drop if the lagged columns were actually created
        lagged_cols_created = [f'feature_{col}_lag1' for col in features_to_lag if f'feature_{col}_lag1' in all_events.columns]
        if lagged_cols_created:
            f1_ready_df = all_events.dropna(subset=lagged_cols_created).copy() # Use .copy() to avoid SettingWithCopyWarning
        else:
            # If no lagged columns were created (e.g., if features_to_lag was empty or columns missing), use the original dataframe (though model likely won't work)
            f1_ready_df = all_events.copy()
            print("Warning: No lagged features were created. Model trained with lagged features will fail.")


        print("✅ Correção aplicada. DataFrame pronto para modelagem.")

        # --- Etapa 4: Definição do Problema e Divisão Temporal dos Dados (CORRIGIDO) (Copied from d9d4daa2) ---
        print("\n🎯 Etapa 4: Definindo problema e dividindo os dados cronologicamente...")
        target = 'Final_Position_Numeric'
        # Ensure feature_cols include all lagged features that were successfully created
        feature_cols = [col for col in f1_ready_df.columns if col.startswith('feature_') and col.endswith('_lag1') and col in lagged_cols_created] + ['Race_Order', 'Team', 'Driver', 'Race_Track']

        # Filter f1_ready_df to only include the columns actually used as features or target
        all_relevant_cols = feature_cols + [target]
        f1_ready_df = f1_ready_df[all_relevant_cols].copy() # Ensure using a copy

        X = f1_ready_df[feature_cols]
        y = f1_ready_df[target]

        # Check if X is empty after filtering and dropping NaNs
        if X.empty:
             print("\n❌ Error: Feature DataFrame (X) is empty after processing. Cannot proceed with evaluation.")
             return None, None, None, None


        test_size = int(len(X) * 0.2)
        X_train, X_test = X[:-test_size], X[-test_size:]
        y_train, y_test = y[:-test_size], y[-test_size:]


        print(f"Dados de Treino: {len(X_train)}, Dados de Teste: {len(X_test)}")

        # Etapa 5: Treinamento dos modelos (Ensure preprocessor is built on X_train columns)
        num_cols = X_train.select_dtypes(include=np.number).columns.tolist()
        cat_cols = X_train.select_dtypes(exclude=np.number).columns.tolist()

        # Ensure columns in preprocessor match columns in X_train
        preprocessor = ColumnTransformer(transformers=[
            ('num', Pipeline(steps=[('imputer', SimpleImputer(strategy="median")), ('scaler', StandardScaler())]), num_cols),
            ('cat', Pipeline(steps=[('imputer', SimpleImputer(strategy="constant", fill_value="missing")), ('onehot', OneHotEncoder(handle_unknown="ignore"))]), cat_cols)
        ])


        # Baseline
        baseline = DummyRegressor(strategy="mean")
        baseline.fit(preprocessor.fit_transform(X_train), y_train) # Fit preprocessor on train data before baseline
        y_pred_base = baseline.predict(preprocessor.transform(X_test)) # Transform test data before predicting

        # Melhor Modelo (with parameters from optimization in cHtlviQShpuq if available, otherwise defaults)
        # Assuming the best parameters found in cHtlviQShpuq were {'model__max_depth': 10, 'model__min_samples_leaf': 4, 'model__n_estimators': 200}
        best_rf_params = {'n_estimators': 200, 'max_depth': 10, 'min_samples_leaf': 4, 'random_state': 42, 'n_jobs': -1}
        # If you want to use the actual best_params from a successful search in cHtlviQShpuq:
        # try:
        #     best_rf_params = search_rf.best_params_['model__'] # Extract params from search result
        # except NameError:
        #      print("Warning: search_rf not found, using default best_rf_params for evaluation.")
        # except KeyError:
        #      print("Warning: Could not extract model params from search_rf.best_params_, using default.")


        best_rf = Pipeline(steps=[('preprocessor', preprocessor),
                                  ('model', RandomForestRegressor(**best_rf_params))]) # Use the defined parameters
        best_rf.fit(X_train, y_train)
        y_pred_final = best_rf.predict(X_test)

        print("\n--- Modelos Treinados with Correct Features and Preprocessing ---")
        return X_test, y_test, y_pred_base, y_pred_final # Return X_test with correct columns

    except FileNotFoundError:
        print("\n❌ CRITICAL ERROR: 'F1_War_Machine.csv' not found. Analysis cannot proceed.")
        return None, None, None, None
    except Exception as e:
        print(f"\n❌ An unexpected error occurred during run_evaluation: {e}")
        # Detailed error tracing for debugging
        import traceback
        print(traceback.format_exc())
        return None, None, None, None


def perform_analysis(X_test, y_test, y_pred_base, y_pred_final):
    # A) Compare baseline vs melhor modelo
    print("\n--- A) Comparativo: Baseline vs. Melhor Modelo ---")
    # Ensure y_test and y_pred_base are not None and have same length
    if y_test is not None and y_pred_base is not None and len(y_test) == len(y_pred_base):
        baseline_metrics = {"MAE": mean_absolute_error(y_test, y_pred_base), "R2": r2_score(y_test, y_pred_base)}
        final_metrics = {"MAE": mean_absolute_error(y_test, y_pred_final), "R2": r2_score(y_test, y_pred_final)}

        comparison_df = pd.DataFrame({'Baseline': baseline_metrics, 'Random Forest Otimizado': final_metrics})
        print(comparison_df)
    else:
        print("Could not compare models due to missing or mismatched prediction results.")


    # B) Análise de Erros
    print("\n--- B) Análise de Erros do Modelo Otimizado ---")
    # Ensure all necessary dataframes/series are available and not empty
    if X_test is not None and y_test is not None and y_pred_final is not None and not X_test.empty:
        results_df = X_test.copy()
        results_df['Posicao_Real'] = y_test.values # Use .values to align with DataFrame index
        results_df['Posicao_Prevista'] = y_pred_final
        results_df['Residuo (Erro)'] = results_df['Posicao_Real'] - results_df['Posicao_Prevista'] # Calculate residue within the combined df

        print("\nCasos com Piores Previsões (Maiores Erros):")
        # Reindex based on the absolute residual values
        worst_predictions = results_df.reindex(results_df['Residuo (Erro)'].abs().sort_values(ascending=False).index)
        # Ensure the columns exist before trying to print them
        cols_to_print = ['Driver', 'Race_Track', 'Posicao_Real', 'Posicao_Prevista', 'Residuo (Erro)']
        # Filter for columns that actually exist in worst_predictions
        existing_cols_to_print = [col for col in cols_to_print if col in worst_predictions.columns]
        if existing_cols_to_print:
             print(worst_predictions[existing_cols_to_print].head(10))
        else:
             print("Required columns for worst predictions table not found in results_df.")


        # Gráfico de Resíduos vs Valores Previstos
        plt.figure(figsize=(10, 6))
        # Ensure x and y data for scatter plot are not empty or None
        if y_pred_final is not None and results_df['Residuo (Erro)'] is not None and len(y_pred_final) == len(results_df['Residuo (Erro)']):
            sns.scatterplot(x=y_pred_final, y=results_df['Residuo (Erro)'], alpha=0.6)
            plt.axhline(y=0, color='r', linestyle='--')
            plt.title('Gráfico de Resíduos (Posição Prevista vs. Erro)', fontsize=16)
            plt.xlabel('Posição Prevista', fontsize=12)
            plt.ylabel('Resíduo (Real - Previsto)', fontsize=12)
            plt.grid(True)
            plt.savefig('residual_plot.png')
            plt.show()
        else:
            print("Could not generate Residuals vs. Predicted plot due to missing or mismatched data.")


        # Gráfico de Resíduos vs. Posição na Corrida Anterior (Need to ensure this lagged feature column exists in X_test)
        lagged_pos_col = 'feature_Final_Position_Numeric_lag1'
        if lagged_pos_col in X_test.columns and results_df['Residuo (Erro)'] is not None and len(X_test) == len(results_df['Residuo (Erro)']):
             plt.figure(figsize=(10, 6))
             sns.scatterplot(x=X_test[lagged_pos_col], y=results_df['Residuo (Erro)'], alpha=0.6)
             plt.axhline(y=0, color='r', linestyle='--')
             plt.title('Gráfico de Resíduos vs. Posição na Corrida Anterior', fontsize=16)
             plt.xlabel('Posição Final na Corrida Anterior (Feature)', fontsize=12)
             plt.ylabel('Resíduo (Real - Previsto)', fontsize=12)
             plt.grid(True)
             plt.savefig('residual_vs_lagged_pos_plot.png')
             plt.show()
        else:
             print(f"Could not generate Residuals vs. {lagged_pos_col} plot because the column was not found in X_test or data is mismatched.")


        # Histogram of Residuals
        if results_df['Residuo (Erro)'] is not None and not results_df['Residuo (Erro)'].empty:
             plt.figure(figsize=(10, 6))
             sns.histplot(results_df['Residuo (Erro)'], kde=True)
             plt.title('Distribuição dos Resíduos', fontsize=16)
             plt.xlabel('Valor do Resíduo', fontsize=12)
             plt.ylabel('Frequência', fontsize=12)
             plt.grid(True)
             plt.savefig('residuals_histogram.png')
             plt.show()
        else:
             print("Could not generate Residuals histogram due to missing or empty residual data.")


    else:
         print("Could not perform error analysis due to missing or empty test data or predictions.")


# --- Execução Principal ---
X_test, y_test, y_pred_base, y_pred_final = run_evaluation()
if X_test is not None and y_test is not None and y_pred_base is not None and y_pred_final is not None:
    perform_analysis(X_test, y_test, y_pred_base, y_pred_final)
else:
    print("\nSkipping performance analysis due to errors or missing data during evaluation.")
````

Para as colunas categóricas:

Em seguida, foi aplicado OneHotEncoder com handle_unknown="ignore" para converter as variáveis categóricas em representações numéricas (one-hot encoding), ignorando categorias desconhecidas que possam aparecer no conjunto de teste.
Estas transformações foram encapsuladas em um ColumnTransformer dentro de um Pipeline para garantir que o pré-processamento seja aplicado de forma consistente e para evitar vazamento de dados entre os conjuntos de treino e teste.

Engenharia de Atributos Detalhada no Projeto F1 War Machine
O objetivo central da sua engenharia de atributos foi contextualizar cada corrida, dando ao modelo uma "memória" do que aconteceu na temporada até aquele momento. Isso foi feito em quatro fases principais:

**Fase 1: Limpeza e Criação de Features Básicas**
Esta fase preparou o terreno, convertendo dados brutos em um formato numérico e consistente que o modelo pode entender.

**Event_Points (Pontos por Evento):**

O que foi feito: A coluna Points original, que continha valores nulos (NaN) para pilotos que não pontuaram, foi convertida na feature Event_Points, onde os NaNs foram substituídos por 0.

Por quê? Modelos de machine learning não conseguem lidar com valores ausentes. Ao preencher com zero, você atribui corretamente a pontuação para todos os pilotos que terminaram fora do top 10, criando uma feature numérica completa e precisa.

**Event_Win (Vitória no Evento):**

O que foi feito: Você criou uma coluna binária (0 ou 1) que recebe o valor 1 se o piloto terminou a corrida (EventType == 'Race') na primeira posição (Position == '1'), e 0 em todos os outros casos.

Por quê? Isso isola o evento mais importante de uma corrida. Em vez de o modelo ter que inferir o que significa Position == '1', você dá a ele uma feature explícita de "vitória". Isso é especialmente útil para rastrear o sucesso de um piloto ao longo da temporada.

**Final_Position_Numeric (Variável Alvo):**

O que foi feito: A coluna Position (ou Final_Position), que é um texto (string), foi convertida em uma coluna numérica. Casos como "DNF" (Did Not Finish) ou outras inconsistências foram tratados e, inteligentemente, substituídos por um valor alto (como 25), efetivamente colocando-os no final do grid.

Por quê? Esta é a criação da sua variável alvo (target). Um modelo de regressão precisa de um número para prever, e esta etapa garante um alvo limpo e numérico para o treinamento.

**Fase 2: Criação de Contexto Temporal**
Aqui você deu ao modelo a noção de "tempo", que é fundamental para a análise.

**Race_Order (Ordem da Corrida):**

O que foi feito: Foi criada uma coluna que numera sequencialmente as corridas de 1 a N para cada temporada.

Por quê? Sem isso, os dados não teriam uma ordem cronológica clara. Race_Order permite que você ordene os eventos corretamente e é a base para as features cumulativas e defasadas.

**Fase 3: Features Cumulativas (A "Memória" da Temporada)**
Esta é a primeira etapa da criação de features avançadas, onde você calcula o estado do campeonato a cada corrida.

**Total_Driver_Points_Season (Pontos Acumulados do Piloto):**

O que foi feito: Para cada piloto, você calculou a soma cumulativa (cumsum) dos Event_Points ao longo da temporada.

Insight: Esta feature representa a classificação do piloto no campeonato a cada instante. Para a 5ª corrida, por exemplo, ela mostra quantos pontos o piloto tinha acumulado nas 4 corridas anteriores.

**Total_Constructor_Points_Season (Pontos Acumulados do Construtor):**

**Fase 4: Features Defasadas (A Correção Preditiva e Anti-Vazamento)**
Esta é a etapa mais sofisticada e metodologicamente crucial do seu projeto. É o que transforma a análise de "o que aconteceu" para "o que provavelmente vai acontecer".

feature_*_lag1 (Features Defasadas):

O que foi feito: Você pegou todas as features cumulativas da Fase 3 e aplicou a função .shift(1).

Por quê (Exemplo Prático):

O Problema (Data Leakage): A feature Total_Driver_Points_Season da 5ª corrida (Miami) contém os pontos incluindo o resultado de Miami. Usar essa informação para prever o resultado de Miami seria "trapacear", pois você estaria usando o resultado para prever ele mesmo.

A Solução (.shift(1)): Ao aplicar o shift, a feature feature_Total_Driver_Points_Season_lag1 para a corrida de Miami conterá o valor de Total_Driver_Points_Season da 4ª corrida (China). Em outras palavras, você está alimentando o modelo com o estado exato do campeonato antes da corrida que você quer prever.

**Insight: Esta etapa é a que torna seu modelo verdadeiramente preditivo. Você está simulando o mundo real, onde para prever uma corrida, só temos acesso a informações do passado. Essa decisão metodológica é o que garante que sua medição de acurácia (R² de 0.18) seja honesta e confiável.**

Se usar DL, descreva: arquitetura, hiperparâmetros, _early stopping_, tamanho do batch, épocas, e se houve **fine-tuning** de modelos pré-treinados.

**Deep Learning para Previsão na F1: A Escolha do Algoritmo
Para um problema como este, onde a sequência de eventos (corridas) é crucial, o algoritmo de Deep Learning mais indicado é o LSTM (Long Short-Term Memory).**

**Por que LSTM?**

**Memória de Longo Prazo**: Uma rede LSTM é um tipo de Rede Neural Recorrente (RNN) projetada para "lembrar" de informações de etapas anteriores na sequência. No contexto da F1, isso é poderoso: o modelo pode aprender que o desempenho de um piloto nas últimas 3 corridas (seu "momentum" ou "forma") é um forte indicador para o resultado da próxima.

**Aprendizagem de Dependências Temporais**: Em vez de você precisar criar manualmente features como "pontos nas últimas X corridas", a LSTM pode aprender essas dependências complexas diretamente dos dados sequenciais. Ela pode identificar se um piloto está em uma trajetória ascendente de performance ou se uma equipe teve uma queda de rendimento após uma atualização.

**O Principal Desafio**:
Redes neurais profundas geralmente precisam de muitos dados para treinar sem superajustar (overfitting). Com um dataset de duas temporadas, o risco de overfitting é real. Portanto, técnicas como Dropout e uma arquitetura de rede simples são essenciais.

**Insights e Diferenças**
Insight de "Momentum" e "Forma":

O que o código faz: A função create_sequences transforma os dados. Em vez de prever uma corrida com base apenas nos dados da corrida anterior, o modelo agora olha para uma janela de 3 corridas (sequence_length=3).

**Insight**: A LSTM aprende o padrão de evolução dentro dessa janela. Ela pode identificar se um piloto terminou em 10º, depois 5º e depois 3º (uma clara trajetória de melhora) e usar essa "tendência" na sua predição. O Random Forest veria cada corrida anterior como um ponto de dado mais independente. A LSTM entende a narrativa da sequência.

**Combate ao Overfitting em Dados Pequenos:**

O que o código faz: As camadas de Dropout(0.2) "desligam" aleatoriamente 20% dos neurônios durante o treinamento.

**Insight**: Isso força a rede a não depender excessivamente de alguns poucos neurônios, tornando-a mais robusta e melhorando sua capacidade de generalizar para dados novos (as corridas futuras). É uma técnica de defesa crucial dado o tamanho limitado do seu dataset.

**Potencial Preditivo para a Batalha McLaren vs. Red Bull:**

Após treinar este modelo, você usaria sua lógica de simulação existente. A diferença é que, para prever a corrida de Singapura, você alimentaria o modelo com a sequência de dados das corridas do Azerbaijão, Itália e Bélgica para cada piloto (Norris, Piastri, Verstappen).

O modelo LSTM pode capturar nuances como "a McLaren teve uma performance excepcional nas últimas duas pistas de alta velocidade" de forma mais eficaz, potencialmente dando uma predição mais precisa para a próxima pista de características semelhantes.

É o caminho para uma "versão 2.0" do F1-War-Machine, se adicionar mais dados históricos (por exemplo, das temporadas de 2022 e 2023) para alimentar a "fome" de dados da LSTM e diminuir ainda mais o risco de overfitting.



````python
# Importações necessárias para Deep Learning usando TensorFlow/Keras
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input
from tensorflow.keras.optimizers import Adam

# --- 1. Carregamento e Preparação dos Dados (similar ao seu código) ---
# Assumindo que 'f1_ready_df' é o seu DataFrame pronto, após o .shift(1)
# Vou recriar uma versão simplificada para o exemplo
try:
    f1_ready_df = pd.read_csv('F1_War_Machine.csv')
    # Simplificação: Focando apenas em corridas e convertendo posições
    f1_ready_df = f1_ready_df[f1_ready_df['EventType'] == 'Race'].copy()
    f1_ready_df['Final_Position_Numeric'] = pd.to_numeric(f1_ready_df['Final_Position'], errors='coerce').fillna(25)
    f1_ready_df.sort_values(by=['Season', 'Race_Order'], inplace=True)
    print("✅ Dados carregados e preparados.")
except FileNotFoundError:
    print("❌ ERRO: Arquivo 'F1_War_Machine.csv' não encontrado.")
    exit()

# --- 2. Seleção de Features e Pré-processamento ---
# Para este exemplo, usaremos apenas features numéricas para simplificar.
# TODO: Você deve incluir suas features categóricas (Team, Driver, Track) com OneHotEncoder
numeric_features = ['Race_Order', 'Event_Points']
target = 'Final_Position_Numeric'

# Criando features de lag (essencial para o contexto da LSTM)
df_model = f1_ready_df[['Driver', 'Season'] + numeric_features + [target]].copy()
for col in numeric_features + [target]:
    # Usamos o lag para representar o estado da corrida anterior
    df_model[f'{col}_lag1'] = df_model.groupby(['Season', 'Driver'])[col].shift(1)

df_model.dropna(inplace=True)

features_final = [f'{col}_lag1' for col in numeric_features + [target]]
X_flat = df_model[features_final]
y_flat = df_model[target]

# Normalização dos dados (crucial para redes neurais)
scaler = StandardScaler()
X_scaled_flat = scaler.fit_transform(X_flat)

# --- 3. CRIAÇÃO DE SEQUÊNCIAS (A MÁGICA DA LSTM) ---
# Vamos transformar os dados em sequências de 'n' corridas passadas para prever a próxima.
def create_sequences(X_data, y_data, sequence_length=3):
    X_seq, y_seq = [], []
    for i in range(len(X_data) - sequence_length):
        X_seq.append(X_data[i:(i + sequence_length)])
        y_seq.append(y_data.iloc[i + sequence_length])
    return np.array(X_seq), np.array(y_seq)

sequence_length = 3 # Usar o histórico das últimas 3 corridas para prever a próxima
X_seq, y_seq = create_sequences(X_scaled_flat, y_flat, sequence_length)

# Divisão em treino e teste (mantendo a ordem cronológica)
test_size = int(len(X_seq) * 0.2)
X_train, X_test = X_seq[:-test_size], X_seq[-test_size:]
y_train, y_test = y_seq[:-test_size], y_seq[-test_size:]

print(f"Formato dos dados de treino (Sequências, Timesteps, Features): {X_train.shape}")

# --- 4. Construção e Treinamento do Modelo LSTM ---
model = Sequential()
model.add(Input(shape=(X_train.shape[1], X_train.shape[2])))
model.add(LSTM(units=50, activation='relu', return_sequences=True)) # Primeira camada LSTM
model.add(Dropout(0.2)) # Dropout para evitar overfitting
model.add(LSTM(units=50, activation='relu')) # Segunda camada LSTM
model.add(Dropout(0.2))
model.add(Dense(units=25, activation='relu')) # Camada densa para processamento
model.add(Dense(units=1, activation='linear')) # Camada de saída para prever a posição

# Compilação do modelo
optimizer = Adam(learning_rate=0.001)
model.compile(optimizer=optimizer, loss='mean_squared_error', metrics=['mean_absolute_error'])
model.summary()

# Treinamento
print("\n🚀 Treinando o modelo LSTM...")
history = model.fit(
    X_train, y_train,
    epochs=50, # Número de épocas de treino
    batch_size=16,
    validation_split=0.1, # Usar parte dos dados de treino para validação
    verbose=1
)

# --- 5. Avaliação ---
print("\n🎯 Avaliando o modelo no conjunto de teste final...")
results = model.evaluate(X_test, y_test, verbose=1)
print(f"Loss no Teste (MSE): {results[0]:.4f}")
print(f"Erro Médio Absoluto (MAE) no Teste: {results[1]:.4f} posições")
````

Quando você executa grid_search.fit(X_train, y_train), o Pipeline garante que o preprocessor (que contém o StandardScaler) aprenda (.fit()) a média e o desvio padrão APENAS dos dados de X_train. Após treinar e otimizar nosso modelo, obtivemos métricas de acurácia realistas. Agora, vamos aprofundar a análise para entender a natureza dos erros do modelo. A melhor forma de fazer isso é através da Análise de Resíduos.

1. O que são Resíduos?
Um resíduo é simplesmente a diferença entre o valor real e o valor previsto pelo modelo para uma única observação.

Fórmula: Resíduo = Posição Final Real - Posição Final Prevista

Interpretação:

Resíduo negativo: O modelo previu uma posição melhor (número menor) do que a real. (Ex: previu 5º, chegou em 8º. Resíduo = 8 - 5 = 3. Correção: A fórmula é Real - Previsto. Ex: previu 8º, chegou em 5º. Resíduo = 5 - 8 = -3. O modelo foi pessimista).

Resíduo positivo: O modelo previu uma posição pior (número maior) do que a real. (Ex: previu 10º, chegou em 6º. Resíduo = 6 - 10 = -4. Correção: Ex: previu 6º, chegou em 10º. Resíduo = 10 - 6 = 4. O modelo foi otimista).

Resíduo próximo de zero: A previsão foi muito acurada.

Analisar a distribuição e os padrões desses erros nos diz se o modelo tem algum viés sistemático.

2. Visualização: Resíduos vs. Valores Previstos
Este é o gráfico mais importante para diagnosticar problemas em um modelo de regressão. Ele nos ajuda a ver se a magnitude do erro está relacionada ao valor da previsão.

O que buscamos? Pontos espalhados aleatoriamente em torno da linha horizontal em y=0, sem nenhum padrão discernível. Isso indica que os erros são consistentes em toda a gama de previsões (homocedasticidade).

Sinais de Problema:

Formato de cone/funil: Se os resíduos se espalham mais para previsões maiores ou menores, indica que o erro do modelo não é constante (heterocedasticidade).

Padrão curvo: Sugere que o modelo não está capturando alguma relação não-linear nos dados.

3. Visualização: Distribuição dos Erros (Histograma)
Este gráfico nos mostra a frequência dos erros de diferentes tamanhos, ajudando a entender se os erros são, em geral, pequenos e centrados em zero.

O que buscamos? Uma distribuição que se assemelhe a uma curva de sino (distribuição normal), com o pico em 0. Isso significa que erros grandes (positivos ou negativos) são raros, e a maioria dos erros é pequena.

Sinais de Problema:

Distribuição assimétrica: Se a cauda for mais longa para um lado, indica que o modelo tem um viés para superestimar ou subestimar as posições.

4. Análise de Viés Sistemático (Resíduos vs. Posição Anterior)
Para investigar se o modelo erra mais em posições específicas, podemos plotar os erros contra uma feature importante, como a posição do piloto na corrida anterior (feature_Final_Position_Numeric_lag1).

Solução de Vazamento de Dados no MVP-F1 Machine Learning
1. O Princípio: Prever o Futuro com Dados do Passado
O objetivo fundamental da correção é garantir que, para prever o resultado de uma corrida, o modelo utilize apenas informações que estavam disponíveis antes daquela corrida começar. O código que você implementou faz exatamente isso ao "empurrar" os dados de uma corrida para a linha da corrida seguinte.

Função Chave: .shift(1)

O que ela faz: Para cada piloto, em cada temporada, a função shift(1) pega o valor de uma coluna (ex: Total_Victories da Corrida 2) e o move para a linha da Corrida 3, dentro de uma nova coluna (feature_Total_Victories_lag1).

Resultado Prático: Na linha da "Corrida 3", a nova feature feature_Total_Victories_lag1 conterá o número de vitórias que o piloto tinha após a "Corrida 2". Esta é a informação correta e justa para se usar na predição da Corrida 3.

2. Isolamento de Dados: A Importância do groupby()
O vazamento de dados poderia ocorrer de formas sutis se não houvesse um isolamento correto dos dados de cada piloto.

Função Chave: .groupby(['Season', 'Driver'])

O que ela faz: Este comando age como uma "parede", instruindo a função .shift(1) a operar de forma independente para cada piloto dentro de cada temporada.

Resultado Prático:

O resultado da última corrida de 2024 de um piloto não vaza como dado inicial para a primeira corrida de 2025.

O desempenho do Piloto A não influencia as features do Piloto B.

Isso garante que a linha do tempo de performance de cada piloto seja única e tratada de forma isolada, como acontece no mundo real.

3. O Ponto de Partida: Tratando a Primeira Corrida
Quando a função .shift(1) é aplicada na primeira corrida da temporada de um piloto, não há uma "corrida anterior" para buscar dados. Isso gera um valor vazio (NaN).

Função Chave: .fillna(0, inplace=True)

O que ela faz: Este comando encontra todos esses valores vazios criados pelo .shift(1) na primeira corrida e os substitui por 0.

Resultado Prático: Isso simula a realidade perfeitamente. Antes da primeira corrida de uma temporada, cada piloto começa com 0 pontos, 0 vitórias, 0 poles, etc. O DataFrame agora reflete corretamente este ponto de partida.

Conclusão: Um DataFrame Pronto para Predições Realistas
Ao final do processo, o seu DataFrame f1_war_machine contém um novo conjunto de colunas (com o prefixo feature_ e sufixo _lag1). São estas colunas que devem ser usadas para treinar o seu modelo de Machine Learning. As colunas originais (como Final_Position, Event_Points) devem ser usadas para definir o seu alvo de predição (variável y), mas nunca como features de entrada (X).

Dessa forma, você garante que seu modelo aprenda a partir de uma simulação justa e realista, aumentando drasticamente a sua capacidade de fazer predições úteis sobre corridas futuras.

````python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from IPython.display import display, HTML
import base64
import warnings
from datetime import datetime

warnings.filterwarnings('ignore')

# --- Tradutor de meses para robustez ---
month_translation = {'janeiro': 'January', 'fevereiro': 'February', 'março': 'March', 'abril': 'April', 'maio': 'May', 'junho': 'June', 'julho': 'July', 'agosto': 'August', 'setembro': 'September', 'outubro': 'October', 'novembro': 'November', 'dezembro': 'December'}

# --- DADOS DO CALENDÁRIO E MAPEAMENTOS ---
calendar_data = [
    (1, 'Grande Prêmio da Austrália', 'Australia', 'Circuito do Grande Prêmio de Melbourne', '16 de março'),(2, 'Grande Prêmio da China', 'China', 'Circuito Internacional de Xangai', '23 de março'),(3, 'Grande Prêmio do Japão', 'Japan', 'Curso Internacional de Corridas de Suzuka', '6 de abril'),(4, 'Grande Prêmio do Barém', 'Bahrain', 'Circuito Internacional do Barém', '13 de abril'),(5, 'Grande Prêmio da Arábia Saudita', 'Saudi Arabia', 'Circuito da Corniche de Gidá', '20 de abril'),(6, 'Grande Prêmio de Miami', 'Miami', 'Autódromo Internacional de Miami', '4 de maio'),(7, 'Grande Prêmio da Emília-Romanha', 'Italy', 'Autódromo Enzo e Dino Ferrari', '18 de maio'),(8, 'Grande Prêmio de Mônaco', 'Monaco', 'Circuito de Mônaco', '25 de maio'),(9, 'Grande Prêmio da Espanha', 'Spain', 'Circuito de Barcelona-Catalunha', '1 de junho'),(10, 'Grande Prêmio do Canadá', 'Canada', 'Circuito Gilles Villeneuve', '15 de junho'),(11, 'Grande Prêmio da Áustria', 'Austria', 'Red Bull Ring', '29 de junho'),(12, 'Grande Prêmio da Grã-Bretanha', 'Great Britain', 'Circuito de Silverstone', '6 de julho'),(13, 'Grande Prêmio da Bélgica', 'Belgium', 'Circuito de Spa-Francorchamps', '27 de julho'),(14, 'Grande Prêmio da Hungria', 'Hungary', 'Hungaroring', '3 de agosto'),(15, 'Grande Prêmio dos Países Baixos', 'Netherlands', 'Circuito de Zandvoort', '31 de agosto'),(16, 'Grande Prêmio da Itália', 'Italy_2', 'Autódromo Nacional de Monza', '7 de setembro'),(17, 'Grande Prêmio do Azerbaijão', 'Azerbaijan', 'Circuito Urbano de Bacu', '21 de setembro'),(18, 'Grande Prêmio de Singapura', 'Singapore', 'Circuito Urbano de Marina Bay', '5 de outubro'),(19, 'Grande Prêmio dos Estados Unidos', 'United States', 'Circuito das Américas', '19 de outubro'),(20, 'Grande Prêmio da Cidade do México', 'Mexico', 'Autódromo Hermanos Rodríguez', '26 de outubro'),(21, 'Grande Prêmio de São Paulo', 'Brazil', 'Autódromo José Carlos Pace', '9 de novembro'),(22, 'Grande Prêmio de Las Vegas', 'Las Vegas', 'Circuito Urbano de Las Vegas', '22 de novembro'),(23, 'Grande Prêmio do Catar', 'Qatar', 'Circuito Internacional de Lusail', '30 de novembro'),(24, 'Grande Prêmio de Abu Dhabi', 'Abu Dhabi', 'Circuito de Yas Marina', '7 de dezembro')
]
calendar_df_info = pd.DataFrame(calendar_data, columns=['Ordem', 'GP', 'Pais', 'Circuito', 'Data'])
calendar_track_order = calendar_df_info['Pais'].tolist()
track_to_country_code = {'Australia': 'au', 'China': 'cn', 'Japan': 'jp', 'Bahrain': 'bh', 'Saudi Arabia': 'sa','Miami': 'us', 'Italy': 'it', 'Monaco': 'mc', 'Spain': 'es', 'Canada': 'ca','Austria': 'at', 'Great Britain': 'gb', 'Belgium': 'be', 'Hungary': 'hu','Netherlands': 'nl', 'Azerbaijan': 'az', 'Singapore': 'sg', 'United States': 'us','Mexico': 'mx', 'Brazil': 'br', 'Las Vegas': 'us', 'Qatar': 'qa', 'Abu Dhabi': 'ae', 'Italy_2': 'it'}

# --- TABELAS DE PONTUAÇÃO ---
points_mapping = {1: 25, 2: 18, 3: 15, 4: 12, 5: 10, 6: 8, 7: 6, 8: 4, 9: 2, 10: 1}
sprint_points_mapping = {1: 8, 2: 7, 3: 6, 4: 5, 5: 4, 6: 3, 7: 2, 8: 1}

# --- 1. Carregamento e Preparação dos Dados ---
print("--- 1. Carregando e Preparando Dados ---")
try:
    df = pd.read_csv('F1_War_Machine.csv')
    print("✅ Arquivo 'F1_War_Machine.csv' carregado com sucesso.")
except FileNotFoundError:
    print("❌ ERRO: Arquivo 'F1_War_Machine.csv' não encontrado."); exit()

df['Team'] = df['Team'].str.strip()
df['Driver'] = df['Driver'].str.strip()
df.rename(columns={'Race_Track': 'Pais'}, inplace=True)
df['Final_Position'] = pd.to_numeric(df['Final_Position'], errors='coerce').fillna(25).astype(int)
df['Pais'] = np.where((df['Race_Order'] == 16) & (df['Pais'] == 'Italy'), 'Italy_2', df['Pais'])
print("✅ Dados limpos e preparados.")

# --- 2. Treinamento do Modelo Preditivo (Omitido por brevidade) ---
print("\n--- 2. Treinando Modelo Preditivo de Posição Final ---")
df_races = df[df['EventType'] == 'Race'].copy(); features_to_drop = ['Season', 'Race_Order', 'EventType', 'Driver', 'Final_Position', 'Event_Points', 'Final_Status', 'Win']; X_train_df = df_races.drop(columns=features_to_drop, errors='ignore'); y_pos = df_races['Final_Position']; numeric_features = X_train_df.select_dtypes(include=np.number).columns.tolist(); categorical_features = X_train_df.select_dtypes(exclude=np.number).columns.tolist(); preprocessor = ColumnTransformer(transformers=[('num', StandardScaler(), numeric_features),('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_features)],remainder='passthrough'); model_pos = Pipeline(steps=[('preprocessor', preprocessor),('regressor', RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1))]); model_pos.fit(X_train_df, y_pos); print("✅ Modelo treinado com sucesso.")

# --- 3. Inicialização dos Pontos Base (Omitido por brevidade) ---
print("\n--- 3. Inicializando Pontos Base até a 17ª Corrida ---")
initial_driver_points = {'Oscar Piastri': 324, 'Lando Norris': 299, 'Max Verstappen': 255, 'George Russell': 212, 'Charles Leclerc': 165, 'Lewis Hamilton': 121, 'Kimi Antonelli': 78, 'Alexander Albon': 70, 'Isack Hadjar': 39, 'Nico Hulkenberg': 37, 'Lance Stroll': 32, 'Carlos Sainz': 31, 'Liam Lawson': 30, 'Fernando Alonso': 30, 'Esteban Ocon': 28, 'Pierre Gasly': 20,  'Yuki Tsunoda': 20, 'Gabriel Bortoleto': 18, 'Oliver Bearman': 16, 'Franco Colapinto': 0, 'Jack Doohan': 0}; initial_constructor_points = {'McLaren Mercedes': 623, 'Mercedes': 290, 'Ferrari': 286, 'Red Bull Racing Honda RBPT': 272, 'Williams Mercedes': 101, 'Racing Bulls Honda RBPT': 72, 'Aston Martin Aramco Mercedes': 62,  'Kick Sauber Ferrari': 55, 'Haas Ferrari': 44, 'Alpine Renault': 20}

# --- 4. Simulação das Corridas Finais (Omitido por brevidade) ---
print("\n--- 4. Executando Simulação das 7 Corridas Finais ---")
drivers_of_interest = list(initial_driver_points.keys()); teams_of_interest = df[['Driver', 'Team']].drop_duplicates().set_index('Driver')['Team'].to_dict(); current_driver_points = initial_driver_points.copy(); current_constructor_points = initial_constructor_points.copy(); final_races_tracks = ['Singapore', 'United States', 'Mexico', 'Brazil', 'Las Vegas', 'Qatar', 'Abu Dhabi']; sprint_races = ['United States', 'Brazil', 'Qatar']; simulation_log = []
for track in final_races_tracks:
    for event_type in ['Sprint', 'Race'] if track in sprint_races else ['Race']:
        for driver in drivers_of_interest:
            team = teams_of_interest.get(driver);
            if not team: continue
            predict_input_data = pd.DataFrame([{'Team': team, 'Pais': track}]); pred_pos = max(1, round(model_pos.predict(predict_input_data)[0])); points_gained = sprint_points_mapping.get(pred_pos, 0) if event_type == 'Sprint' else points_mapping.get(pred_pos, 0); current_driver_points[driver] += points_gained
            if team in current_constructor_points: current_constructor_points[team] += points_gained
            simulation_log.append({'Evento': f'{track} ({event_type})', 'Piloto': driver, 'Equipe': team, 'Pontos no Evento': points_gained, 'Posicao Prevista': pred_pos})
print("✅ Simulação concluída.")

# --- 5. Geração das Planilhas de Resultados com Estilo F1 ---
print("\n--- 5. Gerando Planilhas Elegantes de Resultados ---")
log_df = pd.DataFrame(simulation_log)
font_face_rule = "";
try:
    with open('Formula1 Display-Regular.ttf', 'rb') as f: font_base64 = base64.b64encode(f.read()).decode('utf-8')
    font_face_rule = f"@font-face {{font-family: 'Formula1'; src: url(data:font/truetype;charset=utf-8;base64,{font_base64}) format('truetype');}}"
except FileNotFoundError: print("⚠️ Aviso: Fonte 'Formula1 Display-Regular.ttf' não encontrada.")
base_style = f"""
<style>
    {font_face_rule}
    .f1-table {{ border-collapse: collapse; width: 100%; margin: 25px 0; font-family: 'Formula1', sans-serif; color: #e4e4e4; font-size: 1em; text-align: left; background-color: #15151e; }}
    .f1-table caption {{ font-size: 1.8em; font-weight: bold; color: #ffffff; margin: 15px 0; text-align: left; }}
    .f1-table thead th {{ background-color: #1f1f27; color: #ffffff; padding: 12px 15px; border-bottom: 4px solid #e10600; font-weight: bold; vertical-align: bottom;}}
    .f1-table tbody td {{ padding: 12px 15px; border-bottom: 1px solid #38383f; }}
    .f1-table tbody tr:nth-of-type(even) {{ background-color: #1f1f27; }}
    .f1-table tbody tr:hover {{ background-color: #38383f; }}
</style>
"""
def integer_formatter(val):
    if isinstance(val, (int, float)): return f"{val:.0f}"
    return val

# --- PLANILHA 1: Calendário da Temporada 2025 ---
CURRENT_DATE = datetime(2025, 9, 27)
calendar_df = calendar_df_info.copy()
date_series = calendar_df['Data'].str.lower()
for pt, en in month_translation.items():
    date_series = date_series.str.replace(pt, en)
calendar_df['Data_Completa'] = pd.to_datetime(date_series + ' 2025', format='%d de %B %Y')
calendar_df['Status'] = np.where(calendar_df['Data_Completa'] < CURRENT_DATE, 'Realizada', 'A Realizar')
calendar_df['Bandeira'] = calendar_df['Pais'].map(track_to_country_code).apply(lambda x: f"https://flagcdn.com/w40/{x.split('_')[0]}.png" if x else '')
def render_flag(url): return f'<img src="{url}" width="30">'
def style_status(val): return f'color: {"#2ECC71" if val == "Realizada" else "#95A5A6"}; font-weight: bold;'
styled_calendar = (calendar_df[['Ordem', 'Bandeira', 'GP', 'Circuito', 'Data', 'Status']].style
    .set_table_attributes('class="f1-table"').set_caption("CALENDÁRIO DA TEMPORADA F1-2025")
    .format({'Bandeira': render_flag}).map(style_status, subset=['Status']).hide(axis='index')
)
display(HTML(f"{base_style}{styled_calendar.to_html()}"))

# --- Lógica para criar os Mapas de Resultados ---
real_results_df = df[(df['Season'] == 2025) & (df['EventType'] == 'Race')].pivot_table(index='Driver', columns='Pais', values='Final_Position')
pred_results_df = log_df.pivot_table(index='Piloto', columns='Evento', values='Posicao Prevista')
season_map_pilots = pd.merge(real_results_df, pred_results_df, left_index=True, right_index=True, how='outer')
season_map_pilots['Total Pontos'] = season_map_pilots.index.map(current_driver_points)
season_map_pilots = season_map_pilots.sort_values(by='Total Pontos', ascending=False)
season_map_pilots.insert(0, 'POS', range(1, len(season_map_pilots) + 1)); season_map_pilots.rename_axis('PILOTO', inplace=True); season_map_pilots.reset_index(inplace=True)
driver_to_team_map = df[['Driver', 'Team']].drop_duplicates().set_index('Driver')['Team'].to_dict(); season_map_pilots.insert(2, 'CONSTRUTOR', season_map_pilots['PILOTO'].map(driver_to_team_map))
real_points_constructor = df[df['Season'] == 2025].groupby(['Team', 'Pais'])['Event_Points'].sum().unstack()
pred_points_constructor = log_df.groupby(['Equipe', 'Evento'])['Pontos no Evento'].sum().unstack()
season_map_constructors = pd.merge(real_points_constructor, pred_points_constructor, left_index=True, right_index=True, how='outer')
season_map_constructors['Total Pontos'] = season_map_constructors.index.map(current_constructor_points)
season_map_constructors = season_map_constructors.sort_values(by='Total Pontos', ascending=False)
season_map_constructors.insert(0, 'POS', range(1, len(season_map_constructors) + 1)); season_map_constructors.rename_axis('CONSTRUTOR', inplace=True); season_map_constructors.reset_index(inplace=True)

# --- Geração do HTML com Cabeçalho Customizado e Estilo ---
def create_html_with_flags(df, title, static_cols, track_map, highlight_func=None):
    styler = df.style.format(integer_formatter, na_rep="-").hide(axis='index').set_table_attributes('class="f1-table"')
    if highlight_func:
        styler = styler.apply(highlight_func, axis=1)
    html_body = styler.to_html()

    header_html = "<thead><tr>"
    for col in df.columns:
        header_content = col.replace('_', ' ').upper()
        if col not in static_cols:
            track_name_key = col.split(' (')[0]
            country_code = track_map.get(track_name_key)
            if country_code:
                header_content = f'<div style="text-align:center;"><img src="https://flagcdn.com/w40/{country_code}.png" alt="{track_name_key}" style="height:20px; border: 1px solid #444;"><br><span style="font-size:0.8em;">{header_content}</span></div>'
        header_html += f"<th>{header_content}</th>"
    header_html += "</tr></thead>"

    body_start_index = html_body.find("<tbody>")
    final_html = f"<caption>{title}</caption>" + header_html + html_body[body_start_index:]
    return f'<table class="f1-table">{final_html}</table>'

# --- Ordena colunas e exibe as tabelas FINAIS ---
pred_cols = sorted([c for c in season_map_pilots.columns if any(t in c for t in final_races_tracks)], key=lambda x: (final_races_tracks.index(x.split(' (')[0]), x.find('Sprint')))
final_col_order_pilots = ['POS', 'PILOTO', 'CONSTRUTOR'] + [t for t in calendar_track_order if t in season_map_pilots.columns] + pred_cols + ['Total Pontos']
season_map_pilots_ordered = season_map_pilots.reindex(columns=final_col_order_pilots).fillna("-")
def highlight_drivers(row): return ['background-color: #403d15'] * len(row) if row['PILOTO'] in ['Max Verstappen', 'Lando Norris', 'Oscar Piastri'] else [''] * len(row)
html_pilots = create_html_with_flags(season_map_pilots_ordered, "A) MAPA DE RESULTADOS DA TEMPORADA F1-2025 (PILOTOS)", ['POS', 'PILOTO', 'CONSTRUTOR', 'Total Pontos'], track_to_country_code, highlight_drivers)
display(HTML(f"{base_style}{html_pilots}"))

final_col_order_const = ['POS', 'CONSTRUTOR'] + [t for t in calendar_track_order if t in season_map_constructors.columns] + pred_cols + ['Total Pontos']
season_map_constructors_ordered = season_map_constructors.reindex(columns=final_col_order_const).fillna("-")
def highlight_teams(row): return ['background-color: #403d15'] * len(row) if row['CONSTRUTOR'] in ['McLaren Mercedes', 'Red Bull Racing Honda RBPT'] else [''] * len(row)
html_constructors = create_html_with_flags(season_map_constructors_ordered, "B) MAPA DE RESULTADOS DA TEMPORADA F1-2025 (CONSTRUTORES)", ['POS', 'CONSTRUTOR', 'Total Pontos'], track_to_country_code, highlight_teams)
display(HTML(f"{base_style}{html_constructors}"))

print("\n✅ Novas planilhas de mapa da temporada formatadas foram geradas.")
````
